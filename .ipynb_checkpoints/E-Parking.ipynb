```python
# Import Library
from tkinter import *            # Untuk membuat GUI
from PIL import ImageTk, Image   # Untuk membaca Gambar
from tkinter import ttk          # Untuk mengatur style widget
import pandas as pd              # Untuk membaca dan menyimpan csv
from datetime import datetime    # Untuk menambahkan jam dan tanggal
from tkinter.messagebox import * # Untuk menambahkan notifikasi peringatan

class E_parking:
    
    def __init__(self, master):
        self.window = master
        
        # Atur ukuran jendela menjadi 1280x720 piksel
        self.window.geometry('1280x720') 
        
        # Nonaktifkan kemampuan pengubahan ukuran jendela (0, 0 berarti tidak dapat diubah ukurannya)
        self.window.resizable(0, 0) 
        
        # Memberikan judul pada GUI
        self.window.title('E-Parking') 
            
        def login_page():
            
            def sign_up():
                # Menghapus widget yang ada pada login page, sehingga kembali menjadi sign in page
                self.bgAwal_log.destroy()
                self.welcome.destroy()
                self.notHave.destroy()
                self.signLabel.destroy()
                self.email_label_log.destroy()
                self.email_log_entry.destroy()
                self.pw_label_log.destroy()
                self.pw_log_entry.destroy()
                self.loginButton.destroy()
                self.show_log_button.destroy()
                
                # Menggunakan try karena peringatan email atau sandi salah tidak selalu muncul
                try:
                    self.emailSandi.destroy()
                except:
                    pass

                # Menggunakan try karena peringatan seluruh entry harus terisi tidak selalu muncul
                try:
                    self.wajibMengisi.destroy()
                except:
                    pass
                
                
            # Login Page =========================================================================================
            # Memanggil fungsi yang menghapus peringatan-peringatan   
            destroy_check()
            
            # Background Login
            photo_bgDasar_log = ImageTk.PhotoImage(Image.open('Gambar/Bg dasar.jpg'))
            self.bgAwal_log = Label(self.window, image=photo_bgDasar, width=1280, height=730)
            self.bgAwal_log.image = photo_bgDasar
            self.bgAwal_log.place(x=0, y=0)

            # Login to your account
            self.loginTo = Label(self.window, background="#2c2e43", text="Login to your account", fg="white",
                                 font=("britannic bold", 25), relief=FLAT, width=19, height=1)
            self.loginTo.place(x=123, y=250)

            # Not have account yet?
            self.notHave = Label(self.window, background="#2c2e43", text="Not have account yet?", fg="#B2B1B9",
                              font=("Bahnschrift", 12), relief=FLAT, width=29, height=1)
            self.notHave.place(x=83, y=300)

            # Sign in
            self.signButton = Button(self.window, background="#2c2e43", text="Sign up", fg="#0075F6",
                                  font=("Bahnschrift", 12, "underline"), relief=FLAT, width=5, height=1, command=sign_up)
            self.signButton.place(x=300, y=295)

            # E-mail
            self.email_label_log = Label(self.window, text="E-mail", bg="#2c2e43", fg="#B2B1B9",
                                        font=("yu gothic ui", 12, "bold"))
            self.email_label_log.place(x=133, y=360)
            self.email_log_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                        font=("yu gothic ui ", 12, "bold"), width=47)
            self.email_log_entry.place(x=135, y=390)

            # Password
            self.pw_label_log = Label(self.window, text="Password", bg="#2c2e43", fg="#B2B1B9",
                                        font=("yu gothic ui", 12, "bold"))
            self.pw_label_log.place(x=133, y=430)
            self.pw_log_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                        font=("yu gothic ui ", 12, "bold"), width=47, show="*")
            self.pw_log_entry.place(x=135, y=460)

            # Create button
            login_img = ImageTk.PhotoImage(Image.open('Gambar/login.png'))
            self.loginButton = Button(self.window, image=login_img,
                                       relief=FLAT, height=40, width=435, bg="#2c2e43", activebackground="#2c2e43",
                                      command=self.homePage)
            self.loginButton.image = login_img
            self.loginButton.place(x=125, y= 520)
            
            # Fungsi untuk menampilkan password pada entry
            def show():
                self.show_log_button.configure(image=self.show_image, command=hide)
                self.pw_log_entry.config(show='')

            # Fungsi untuk sembunyikan password pada entry
            def hide():
                self.show_log_button.configure(image=self.hide_image, command=show)
                self.pw_log_entry.config(show='*')

            # tombol show / hide
            self.show_log_button = Button(self.window, image=self.hide_image, command=show, relief=FLAT,
                                    activebackground="#5d5f70"
                                    , borderwidth=0, background="#5d5f70", cursor="hand2")
            self.show_log_button.image = self.show_image
            self.show_log_button.place(x=540, y=462)
            
        
        # Fungsi untuk menghapus peringatan-peringatan ============================================================
        def destroy_check():
                    try:
                        self.wajibMengisi.destroy()
                    except:
                        pass

                    try:
                        self.wajibPw.destroy()
                    except:
                        pass
                    
                    try:
                        self.wajibEmail.destroy()
                    except:
                        pass
                    try:
                        self.emailUsed.destroy()
                    except:
                        pass
                    
        # Fungsi untuk memeriksa apakah entry yang telah di isi sudah benar =========================================================
        def checking():
            # Kondisi yg mewajibkan mengisi seluruh kolom
            if self.namaDepan_entry.get() and self.namaBelakang_entry.get() and self.email_entry.get() and self.role_box.get() and self.pw_entry.get() and self.confirm_pw_entry.get():

                # Kondisi yg mewajibkan password & konfirmasi password sama
                if self.pw_entry.get() == self.confirm_pw_entry.get():
                        
                    # Kondisi yg mewajibkan email mengandung @
                    if "@" in self.email_entry.get():
                        
                        # membaca csv
                        dataset = pd.read_csv('E-parking.csv')

                        # Membuat list seluruh email yg ada di csv
                        email = [i for i in dataset.email]

                        # Kondisi untuk mengecek email sudah dipakai / belum  
                        if self.email_entry.get() in email:
                            # Fungsi untuk menghapus peringatan
                            destroy_check()
                            
                            # Peringatan email sudah digunakan
                            self.emailUsed = Label(self.window, background="#2c2e43", text='Email has been used',
                                        fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                            self.emailUsed.place(x=210, y=540)
                            
                        # Ketika email belum digunakan
                        else:
                            # Fungsi untuk menghapus peringatan
                            destroy_check()
                            
                            # Menyimpan data register
                            dataset.loc[len(dataset.index)] = [self.namaDepan_entry.get(), self.namaBelakang_entry.get(), self.email_entry.get(), self.role_box.get(), self.pw_entry.get(), 0, "", ""]
                            dataset.to_csv("E-parking.csv", index=False, header=True)
                            
                            # Memanggil halaman login
                            login_page()
                

                    # Jika email tidak mengandung simbol @
                    else:
                        # Fungsi untuk menghapus peringatan
                        destroy_check()
                        
                        # Peringatan tulis email dengan benar
                        self.wajibEmail = Label(self.window, background="#2c2e43", text='Write e-mail correctly',
                            fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                        self.wajibEmail.place(x=210, y=540)

                # Jika konfirmasi password tidak sama
                else:
                    # Fungsi untuk menghapus peringatan
                    destroy_check()
                    
                    # Peringatan untuk menulis password dengan benar
                    self.wajibPw = Label(self.window, background="#2c2e43", text='Write the password correctly',
                        fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                    self.wajibPw.place(x=220, y=540)


            # Jika kolom tidak semua terisi
            else:
                destroy_check()
                self.wajibMengisi = Label(self.window, background="#2c2e43", text='All inputs are required',
                            fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                self.wajibMengisi.place(x=200, y=540)

            
        # Background
        photo_bgDasar = ImageTk.PhotoImage(Image.open('Gambar/Bg dasar.jpg'))
        self.bgAwal = Label(self.window, image=photo_bgDasar, width=1280, height=730)
        self.bgAwal.image = photo_bgDasar
        self.bgAwal.place(x=0, y=0)
        
        # Start for free
        self.freeLabel = Label(self.window, background="#2c2e43", text="Start For Free", fg="#B2B1B9",
                              font=("Bahnschrift", 12), relief=FLAT, width=19, height=1)
        self.freeLabel.place(x=99, y=200)
        
        # Create New Account
        self.createLabel = Label(self.window, background="#2c2e43", text="Create New Account", fg="white",
                                 font=("britannic bold", 25), relief=FLAT, width=19, height=1)
        self.createLabel.place(x=105, y=230)
        
        # Already Have an account
        self.alreadyLabel = Label(self.window, background="#2c2e43", text="Already have an account?", fg="#B2B1B9",
                              font=("Bahnschrift", 12), relief=FLAT, width=29, height=1)
        self.alreadyLabel.place(x=95, y=280)
        
        # Log in
        self.alreadyLabel = Button(self.window, background="#2c2e43", text="Log in", fg="#0075F6",
                              font=("Bahnschrift", 12, "underline"), relief=FLAT, width=5, height=1, command=login_page)
        self.alreadyLabel.place(x=321, y=275)
        
        # First name
        self.namaDepan_label = Label(self.window, text="First Name", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.namaDepan_label.place(x=133, y=340)
        self.namaDepan_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                    font=("yu gothic ui ", 12, "bold"))
        self.namaDepan_entry.place(x=135, y=370, width=200)
        
        # Last name
        self.namaBelakang_label = Label(self.window, text="Last Name", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.namaBelakang_label.place(x=370, y=340)
        self.namaBelakang_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                    font=("yu gothic ui ", 12, "bold"))
        self.namaBelakang_entry.place(x=370, y=370, width=200)
        
        # E-mail
        self.email_label = Label(self.window, text="E-mail", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.email_label.place(x=133, y=410)
        self.email_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                    font=("yu gothic ui ", 12, "bold"))
        self.email_entry.place(x=135, y=440, width=200)
        
        # Mengganti warna combo box
        style= ttk.Style()
        style.theme_use('clam')
        style.configure("TCombobox", fieldbackground= "#5d5f70", background= "#5d5f70")
        
        # Role
        self.role = Label(self.window, text="Role", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.role.place(x=370, y=410)
        self.list_role = ["Pengguna", "Admin"] # List pilihan dalam combobox
        self.role_box = ttk.Combobox(self.window, values=self.list_role, style="TCombobox", background="#2c2e43")
        self.role_box.place(x=372, y=440, height=23 ,width=200)
        
        # Password
        self.pw_label = Label(self.window, text="Password", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.pw_label.place(x=133, y=480)
        self.pw_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                    font=("yu gothic ui ", 12, "bold"), show="*")
        self.pw_entry.place(x=135, y=510, width=200)
        
        # Konfirmasi password
        self.confirm_pw_label = Label(self.window, text="Confirm Password", bg="#2c2e43", fg="#B2B1B9",
                                    font=("yu gothic ui", 12, "bold"))
        self.confirm_pw_label.place(x=370, y=480)
        self.confirm_pw_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                    font=("yu gothic ui ", 12, "bold"), show="*")
        self.confirm_pw_entry.place(x=370, y=510, width=200)
        
        # Create button
        create_img = ImageTk.PhotoImage(Image.open('Gambar/Create Account.png'))
        self.createButton = Button(self.window, image=create_img,
                                   relief=FLAT, height=40, width=435, bg="#2c2e43", activebackground="#2c2e43",
                                  command=checking)
        self.createButton.image = create_img
        self.createButton.place(x=127, y= 570)
        
        # Fungsi untuk menunjukkan password pada entry yang di isi
        def show():
            self.show_button.configure(image=self.show_image, command=hide)
            self.pw_entry.config(show='')
            self.confirm_pw_entry.config(show='')

        # Fungsi untuk menyembunyikan password pada entry yang di isi
        def hide():
            self.show_button.configure(image=self.hide_image, command=show)
            self.pw_entry.config(show='*')
            self.confirm_pw_entry.config(show='*')
            
        #gambar Show & hide
        self.show_image = ImageTk.PhotoImage(Image.open('gambar/show.png'))
        self.hide_image = ImageTk.PhotoImage(Image.open('gambar/hide.png'))

        # tombol show / hide
        self.show_button = Button(self.window, image=self.hide_image, command=show, relief=FLAT,
                                activebackground="#5d5f70"
                                , borderwidth=0, background="#5d5f70", cursor="hand2")
        self.show_button.image = self.show_image
        self.show_button.place(x=315, y=512)
        

    # Halaman home ========================================================================================
    def homePage(self):
        # Fungsi untuk menghapus peringatan-peringatan
        def destroy_check_log():
            # Menggunakan try karena peringatan email/sandi salah tidak selalu muncul
            try:
                self.emailSandi.destroy()
            except:
                pass

            # Menggunakan try karena peringatan wajib mengisih seluruh entry tidak selalu muncul
            try:
                self.wajibMengisi.destroy()
            except:
                pass


        # wajib mengisi seluruh entry
        if self.email_log_entry.get() and self.pw_log_entry.get():
            
            # Membaca dataset
            dataset = pd.read_csv('E-parking.csv')
            
            # Membuat list yg berisi semua email
            email = [i for i in dataset.email]
            
            # Mengurutkan dataset menggunakan bubble sort berdasarkan kolom email
            dataset = self.bubble_sort(dataset, 'email')
            
            # Mereset indeks dataset agar dimulai dari 0 dan menggantikan dataset awal dengan dataset yang telah direset
            dataset.reset_index(drop=True, inplace=True)
            
            # Menyimpan dataset
            dataset.to_csv("E-parking.csv", index=False, header=True)
            
            # Menlakukan pencarian email yang di inputkan menggunakan binary search pada dataset e-parking
            self.idx = self.binary_search("E-parking.csv", self.email_log_entry.get(), "email")

            # Kondisi untuk cek email sesuai dengan email di dataset
            if self.email_log_entry.get() in email:
                
                # Kondisi untuk cek password yg di input sama dgn password dataset
                if self.pw_log_entry.get() == dataset.password[self.idx]:
                    
                    # Kondisi apabila role sama dgn pengguna
                    if dataset.role[self.idx] == "Pengguna":
                    
                        # Fungsi untuk mengembalikan tombol pada pojok kiri menjadi warna abu abu dan menambahkan beberapa label
                        def change_color():
                            # Nama pojok kanan atas
                            self.nickname = Label(self.window, text=f"Hi, {dataset['first name'][self.idx]}", bg="#24273a", fg="white",
                                            font=("yu gothic ui", 12, "bold"), width=50, anchor="e")
                            self.nickname.place(x=740, y=30)

                            # Home button
                            home_img = ImageTk.PhotoImage(Image.open('Gambar/Home button.jpg'))
                            self.homeButton = Button(self.window, image=home_img,
                                                       relief=FLAT, height=30, width=100, bg="#2c2e43", activebackground="#2c2e43", command=self.homePage)
                            self.homeButton.image = home_img
                            self.homeButton.place(x=40, y=160)
                            
                            # Start Parking button
                            home_img = ImageTk.PhotoImage(Image.open('Gambar/start parkir icon1.jpg'))
                            self.homeButton = Button(self.window, image=home_img,
                                                       relief=FLAT, height=30, width=150, bg="#2c2e43", activebackground="#2c2e43", command=startParking)
                            self.homeButton.image = home_img
                            self.homeButton.place(x=30, y=210)

                            # Find my car button kiri
                            car_img = ImageTk.PhotoImage(Image.open('Gambar/mobil1.jpg'))
                            self.findButton = Button(self.window, image=car_img,
                                                       relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", command=findMycar)
                            self.findButton.image = car_img
                            self.findButton.place(x=35, y=260)

                            # Button transaksi kiri
                            dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi1.jpg'))
                            self.payButton = Button(self.window, image=dollar_img,
                                                       relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43", command=transaction)
                            self.payButton.image = dollar_img
                            self.payButton.place(x=36, y=315)



                        # Start parking=======================================================================================     
                        def startParking():
                            # Memuat dataset
                            dataset = pd.read_csv('E-parking.csv')

                            # Fungsi pada tombol done
                            def done(): 
                                # Jika tidak ada tempat parkir yang dipilih
                                if self.selected_seat == None:
                                    showinfo("Notification", "Anda belum memilih tempat parkir")
                                
                                # jika sudah memilih tempat parkir
                                else:
                                    # Menyimpan lokasi parkir dan waktu mulai
                                    dataset.loc[self.idx,['lokasi_parkir', 'waktu_mulai']] = [self.value, datetime.now().strftime('%Y-%b-%d %H:%M:%S')]
                                    
                                    # Menjadikan lokasi parkir yang dipilih tidak bisa di pencet lagi
                                    self.selected_seat.config(state=DISABLED)
                                    
                                    # Menjadikan tombol done tidak bisa dipencet lagi
                                    self.doneButton.config(state=DISABLED)
                                    
                                    # Me-reset variabel selected seat menjadi kosong
                                    self.selected_seat = None

                                    # Menyimpan datasaet
                                    dataset.to_csv("E-parking.csv", index=False, header=True)
                                    
                                    # Membuat varabel yang berisi seluruh lokasi parkir pada dataset
                                    self.ketersediaan_parkir = [i for i in dataset.lokasi_parkir]
                                    
                                    # Notifikasi tempat parkir sukses dipesan
                                    showinfo("Notification", "Tempat parkir sukses dipesan, silahkan parkir ditempat yang telah dipilih")
                                    
                                    # Mengrangi jumlah seluruh tempat parkir dengan tempat parkir yang telah digunakan
                                    self.seluruhParkir = set(self.spaceParkir) - set(self.ketersediaan_parkir)
                                    
                                    # Update teks tempat parkir terdekat
                                    self.tempatDekat.configure(text=f"{self.dict_parkir[list(self.seluruhParkir)[0]]} dan {self.dict_parkir[list(self.seluruhParkir)[1]]}")

                                    
                            # Fungsi untuk menuju ke lantai 1
                            def lt1():

                                # Memanggil fungsi untuk meletakkan kotak tempat parkir
                                tempatParkir(350, 350, "b1", "B1", 6, "b2", "B2", 7, "b3", "B3", 8, "b4", "B4", 9, "b5", "B5", 10)
                                tempatParkir(600, 350, "c1", "C1", 11, "c2", "C2", 12, "c3", "C3", 13, "c4", "C4", 14, "c5", "C5", 15)
                                tempatParkir(410, 530, "a1", "A1", 1, "a2", "A2", 2, "a3", "A3", 3, "a4", "A4", 4, "a5", "A5", 5)
                                tempatParkir(660, 530, "d1", "D1", 16, "d2", "D2", 17, "d3", "D3", 18, "d4", "D4", 19, "d5", "D5", 20)

                                # Menganti teks dan command pada tombol pindah lantai
                                self.ltKiri.configure(text="Lt2", command=lt2)
                                self.ltKanan.configure(text="Lt3", command=lt3)

                            def lt2():
                                # Mengganti background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/start park2 bg.jpg'))
                                self.bgHome.configure(image=photo_bgHome)
                                self.bgHome.image = photo_bgHome

                                # Memanggil fungsi untuk meletakkan kotak tempat parkir
                                tempatParkir(410, 530, "e1", "E1", 21, "e2", "E2", 22, "e3", "E3", 23, "e4", "E4", 24, "e5", "E5", 25)
                                tempatParkir(350, 350, "f1", "F1", 26, "f2", "F2", 27, "f3", "F3", 28, "f4", "F4", 29, "f5", "F5", 30)
                                tempatParkir(600, 350, "g1", "G1", 31, "g2", "G2", 32, "g3", "G3", 33, "g4", "G4", 34, "g5", "G5", 35)
                                tempatParkir(660, 530, "h1", "H1", 36, "h2", "H2", 37, "h3", "H3", 38, "h4", "H4", 39, "h5", "H5", 40)

                                # Menganti teks dan command pada tombol pindah lantai
                                self.ltKiri.configure(text="Lt1", command=lt1)
                                self.ltKanan.configure(text="Lt3", command=lt3)

                            def lt3():
                                # Mengganti background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/start park3 bg.jpg'))
                                self.bgHome.configure(image=photo_bgHome)
                                self.bgHome.image = photo_bgHome

                                # Memanggil fungsi untuk meletakkan kotak tempat parkir
                                tempatParkir(410, 530, "i1", "I1", 41, "i2", "I2", 42, "i3", "I3", 43, "i4", "I4", 44, "i5", "I5", 45)
                                tempatParkir(350, 350, "j1", "J1", 46, "j2", "J2", 47, "j3", "J3", 48, "j4", "J4", 49, "j5", "J5", 50)
                                tempatParkir(600, 350, "k1", "K1", 51, "k2", "K2", 52, "k3", "K3", 53, "k4", "K4", 54, "k5", "K5", 55)
                                tempatParkir(660, 530, "l1", "L1", 56, "l2", "L2", 57, "l3", "L3", 58, "l4", "L4", 59, "l5", "L5", 60)

                                # Menganti teks dan command pada tombol pindah lantai
                                self.ltKiri.configure(text="Lt1", command=lt1)
                                self.ltKanan.configure(text="Lt2", command=lt2)


                            # fugsi kotak kotak parkir
                            def tempatParkir(locx, locy, namaVar1, teks1, value1, namaVar2, teks2, value2, namaVar3, teks3, value3, namaVar4, teks4, value4, namaVar5, teks5, value5):
                                # Membuat list yang berisi tempat parkir yang telah digunakan
                                self.parkirDigunakan = [i for i in dataset.lokasi_parkir]

                                # Membuat tombol pertama
                                setattr(self, namaVar1, Button(self.window, text=teks1, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                               activebackground="#4a495b"))
                                getattr(self, namaVar1).place(x=locx, y=locy)

                                # Mengubah warna tombol pertama jika tempat parkir telah digunakan
                                if value1 in self.parkirDigunakan:
                                    getattr(self, namaVar1)["bg"] = "#4a495b"
                                    getattr(self, namaVar1)["state"] = DISABLED
                                else:
                                    # Mengatur perintah saat tombol pertama ditekan
                                    getattr(self, namaVar1)["command"] = lambda: change_button_color(getattr(self, namaVar1), value1)

                                # Membuat tombol kedua
                                setattr(self, namaVar2, Button(self.window, text=teks2, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                               activebackground="#4a495b"))
                                getattr(self, namaVar2).place(x=locx+35, y=locy)

                                # Mengubah warna tombol kedua jika tempat parkir telah digunakan
                                if value2 in self.parkirDigunakan:
                                    getattr(self, namaVar2)["bg"] = "#4a495b"
                                    getattr(self, namaVar2)["state"] = DISABLED
                                else:
                                    # Mengatur perintah saat tombol kedua ditekan
                                    getattr(self, namaVar2)["command"] = lambda: change_button_color(getattr(self, namaVar2), value2)

                                # Membuat tombol ketiga
                                setattr(self, namaVar3, Button(self.window, text=teks3, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                               activebackground="#4a495b"))
                                getattr(self, namaVar3).place(x=locx+70, y=locy)

                                # Mengubah warna tombol ketiga jika tempat parkir telah digunakan
                                if value3 in self.parkirDigunakan:
                                    getattr(self, namaVar3)["bg"] = "#4a495b"
                                    getattr(self, namaVar3)["state"] = DISABLED
                                else:
                                    # Mengatur perintah saat tombol ketiga ditekan
                                    getattr(self, namaVar3)["command"] = lambda: change_button_color(getattr(self, namaVar3), value3)

                                # Membuat tombol keempat
                                setattr(self, namaVar4, Button(self.window, text=teks4, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                               activebackground="#4a495b"))
                                getattr(self, namaVar4).place(x=locx+105, y=locy)

                                # Mengubah warna tombol keempat jika tempat parkir telah digunakan
                                if value4 in self.parkirDigunakan:
                                    getattr(self, namaVar4)["bg"] = "#4a495b"
                                    getattr(self, namaVar4)["state"] = DISABLED
                                else:
                                    # Mengatur perintah saat tombol keempat ditekan
                                    getattr(self, namaVar4)["command"] = lambda: change_button_color(getattr(self, namaVar4), value4)

                                # Membuat tombol kelima
                                setattr(self, namaVar5, Button(self.window, text=teks5, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                               activebackground="#4a495b"))
                                getattr(self, namaVar5).place(x=locx+140, y=locy)

                                # Mengubah warna tombol kelima jika tempat parkir telah digunakan
                                if value5 in self.parkirDigunakan:
                                    getattr(self, namaVar5)["bg"] = "#4a495b"
                                    getattr(self, namaVar5)["state"] = DISABLED
                                else:
                                    # Mengatur perintah saat tombol kelima ditekan
                                    getattr(self, namaVar5)["command"] = lambda: change_button_color(getattr(self, namaVar5), value5)


                            # Start Parking ===========================================================================================================
                            # Backgrougnd
                            photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/start park bg.jpg'))
                            self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                            self.bgHome.image = photo_bgHome
                            self.bgHome.place(x=0, y=0)

                            done_img = ImageTk.PhotoImage(Image.open('Gambar/done.png'))
                            self.doneButton = Button(self.window, image=done_img,
                                                       relief=FLAT, height=40, width=100, bg="#2c2e43", activebackground="#2c2e43",
                                                      command=done)
                            self.doneButton.image =done_img
                            self.doneButton.place(x=565, y= 610)

                            # Jika sudah melakukan parkir sebelumnya
                            if dataset.lokasi_parkir[self.idx] != 0:
                                
                                # Tombol done tidak bisa ditekan
                                self.doneButton["state"] = DISABLED
                                
                                # Peringatan anda sudah melakukan parkir
                                self.sudahParkir = Label(self.window, text=f"Anda sudah melakukan parkir", font=("yu gothic ui", 12), bg="#2c2e43", fg="red", justify='center', anchor='center')
                                self.sudahParkir.place(x=515, y=590)

                            # Memangil fungsi untuk merubah warna tombol pada sisi kiri agar menjadi abu abu
                            change_color()

                            # Start Parking button
                            home_img = ImageTk.PhotoImage(Image.open('Gambar/start parkir icon.jpg'))
                            self.homeButton = Button(self.window, image=home_img,
                                                       relief=FLAT, height=30, width=150, bg="#2c2e43", activebackground="#2c2e43", command=startParking)
                            self.homeButton.image = home_img
                            self.homeButton.place(x=30, y=210)

                            # waktu dan jam
                            self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.datetime_label.place(x=1007, y=160)
                            # Memanggil fungsi untuk update jam dan tanggal
                            self.update_datetime()

                            # label tanggal
                            self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                            self.date_label.place(x=1007, y=140)

                            # Label waktu
                            self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                            self.time_label.place(x=1154, y=140)

                            # Biodata akun di pojok kanan bawah
                            self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioFname.place(x=1100, y=503)

                            self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioLname.place(x=1100, y=546)

                            self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                            self.bioEmail.place(x=1100, y=583)

                            self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioRole.place(x=1100, y=622)

                            # kotak parkiran
                            self.ruangParkir = []
                            
                            # Varuabel yang menunjukkan apakah ada kotak yang dipilih
                            self.selected_seat = None

                            # Fungsi agar hanya satu kotak yang bisa dipilih
                            def change_button_color(button, item):
                                # Jika ada kotak yang dipilih
                                if self.selected_seat:
                                    # Jika kotak yang dipilih = kotak yang di klik saat ini
                                    if self.selected_seat == button:
                                        # Mengembalikan warna button menjadi kuning semula
                                        button.config(bg="#f4ce21")
                                        # Me reset variabel menjadi kosong kembali
                                        self.selected_seat = None
                                        return
                                    # Mengganti warna tombol selected_sear menjadi kuning semula
                                    self.selected_seat.config(bg="#f4ce21")

                                # Mengganti warna button menjadi abu abu
                                button.config(bg="#4a495b")
                                # Button yang di klik menjadi selectead_seat
                                self.selected_seat = button
                                # Mendefinisikan value berdasarkan parameter item
                                self.value = item

                            # Tombol lantai kiri
                            self.ltKiri = Button(self.window, text="Lt.2", relief=FLAT, height=1, width=2, bg="#f4ce21",
                                               fg="white", activebackground="#4a495b", command=lt2)
                            self.ltKiri.place(x=720, y=620)

                            # Tombol lantai kanan
                            self.ltKanan = Button(self.window, text="Lt.3", relief=FLAT, height=1, width=2, bg="#f4ce21",
                                               fg="white", activebackground="#4a495b", command=lt3)
                            self.ltKanan.place(x=760, y=620)

                            # Memanggil fungsi lt1
                            self.dict_parkir = {
                                                1: 'A1', 2: 'A2', 3: 'A3', 4: 'A4', 5: 'A5',
                                                6: 'B1', 7: 'B2', 8: 'B3', 9: 'B4', 10: 'B5',
                                                11: 'C1', 12: 'C2', 13: 'C3', 14: 'C4', 15: 'C5',
                                                16: 'D1', 17: 'D2', 18: 'D3', 19: 'D4', 20: 'D5',
                                                21: 'E1', 22: 'E2', 23: 'E3', 24: 'E4', 25: 'E5',
                                                26: 'F1', 27: 'F2', 28: 'F3', 29: 'F4', 30: 'F5',
                                                31: 'G1', 32: 'G2', 33: 'G3', 34: 'G4', 35: 'G5',
                                                36: 'H1', 37: 'H2', 38: 'H3', 39: 'H4', 40: 'H5',
                                                41: 'I1', 42: 'I2', 43: 'I3', 44: 'I4', 45: 'I5',
                                                46: 'J1', 47: 'J2', 48: 'J3', 49: 'J4', 50: 'J5',
                                                51: 'K1', 52: 'K2', 53: 'K3', 54: 'K4', 55: 'K5',
                                                56: 'L1', 57: 'L2', 58: 'L3', 59: 'L4', 60: 'L5'
                                                }

                            lt1()

                            # Membuat list seluruh tempat parkir
                            self.spaceParkir = [i for i in range(1,61)]
                            
                            # Membuat list yang berisi tempat parkir yang telah dipakai
                            self.ketersediaan_parkir = [i for i in dataset.lokasi_parkir]
                            
                            # Mencari parkir yang tersedia
                            self.seluruhParkir = set(self.spaceParkir) - set(self.ketersediaan_parkir)

                            # Label tempat parkir terdekat
                            self.dekat = Label(self.window, text=f'Tempat parkir terdekat adalah :\n', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                            self.dekat.place(x=1003, y=200)

                            # Label nama parkir terdekat
                            self.tempatDekat = Label(self.window, text=f'{self.dict_parkir[list(self.seluruhParkir)[0]]} dan {self.dict_parkir[list(self.seluruhParkir)[1]]}', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.tempatDekat.place(x=1070, y=221)


                        # Find my car ==================================================================================================================================
                        def findMycar():
                            # Membaca dataset
                            dataset = pd.read_csv("E-parking.csv")

                            # Jika belum parkir
                            if dataset.lokasi_parkir[self.idx] == 0:
                                showinfo("Notification", "Maaf, anda belum parkir\nsilahkan memilih tempat parkir terlebih dahulu")
                            else:
                                # Background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car.jpg'))
                                self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                                self.bgHome.image = photo_bgHome
                                self.bgHome.place(x=0, y=0)

                                # Fungsi untuk mengganti warna tombol sisi kiri menjadi abu abu
                                change_color()

                                # Find my car button
                                car_img = ImageTk.PhotoImage(Image.open('Gambar/mobil.jpg'))
                                self.findButton = Button(self.window, image=car_img,
                                                           relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", command=findMycar)
                                self.findButton.image = car_img
                                self.findButton.place(x=35, y=260)

                                # waktu dan jam
                                self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.datetime_label.place(x=1007, y=160)
                                self.update_datetime()

                                # label tanggal
                                self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.date_label.place(x=1007, y=140)

                                # Label waktu
                                self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.time_label.place(x=1154, y=140)

                                # Biodata akun pojok kanan bawah
                                self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioFname.place(x=1100, y=503)
                                
                                self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioLname.place(x=1100, y=546)

                                self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                                self.bioEmail.place(x=1100, y=583)

                                self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioRole.place(x=1100, y=622)

                                # Kotak kotak
                                def tempatParkir(locx, locy, namaVar1, teks1, value1, namaVar2, teks2, value2, namaVar3, teks3, value3, namaVar4, teks4, value4, namaVar5, teks5, value5):

                                    # Memeriksa tempat parkir yang sudah digunakan
                                    self.ketersediaan_parkir = [i for i in dataset.lokasi_parkir]

                                    # Membuat dan menempatkan tombol pertama
                                    setattr(self, namaVar1, Button(self.window, text=teks1, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar1).place(x=locx, y=locy)

                                    # Memeriksa apakah tempat parkir pertama sudah digunakan
                                    if value1 in self.ketersediaan_parkir:
                                        # Jika value 1 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx] == value1:
                                            getattr(self, namaVar1)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar1)["bg"] = "#4a495b"
                                            getattr(self, namaVar1)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol kedua
                                    setattr(self, namaVar2, Button(self.window, text=teks2, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar2).place(x=locx+35, y=locy)

                                    # Memeriksa apakah tempat parkir kedua sudah digunakan
                                    if value2 in self.ketersediaan_parkir:
                                        # Jika value 2 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx] == value2:
                                            getattr(self, namaVar2)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar2)["bg"] = "#4a495b"
                                            getattr(self, namaVar2)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol ketiga
                                    setattr(self, namaVar3, Button(self.window, text=teks3, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar3).place(x=locx+70, y=locy)

                                    # Memeriksa apakah tempat parkir ketiga sudah digunakan
                                    if value3 in self.ketersediaan_parkir:
                                        # Jika value 3 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx] == value3:
                                            getattr(self, namaVar3)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar3)["bg"] = "#4a495b"
                                            getattr(self, namaVar3)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol keempat
                                    setattr(self, namaVar4, Button(self.window, text=teks4, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar4).place(x=locx+105, y=locy)

                                    # Memeriksa apakah tempat parkir keempat sudah digunakan
                                    if value4 in self.ketersediaan_parkir:
                                        # Jika value 4 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx] == value4:
                                            getattr(self, namaVar4)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar4)["bg"] = "#4a495b"
                                            getattr(self, namaVar4)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol kelima
                                    setattr(self, namaVar5, Button(self.window, text=teks5, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar5).place(x=locx+140, y=locy)

                                    # Memeriksa apakah tempat parkir keempat sudah digunakan
                                    if value5 in self.ketersediaan_parkir:
                                        # Jika value 5 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx] == value5:
                                            getattr(self, namaVar5)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar5)["bg"] = "#4a495b"
                                            getattr(self, namaVar5)["state"] = DISABLED    

                                # Jika lokasi parkir pengguna <= 20 maka yang dipanggil adalah halaman lt1
                                if dataset.lokasi_parkir[self.idx] <= 20:
                                    # Daftar nama kotak parkir di l1
                                    self.dict_parkir = {1: 'A1', 2: 'A2', 3: 'A3', 4: 'A4', 5: 'A5',
                                                     6: 'B1', 7: 'B2', 8: 'B3', 9: 'B4', 10: 'B5',
                                                     11: 'C1', 12: 'C2', 13: 'C3', 14: 'C4', 15: 'C5',
                                                     16: 'D1', 17: 'D2', 18: 'D3', 19: 'D4', 20: 'D5'}

                                    # Fungsi untuk memanggil kotak parkir lt1
                                    tempatParkir(410, 530, "a1", "A1", 1, "a2", "A2", 2, "a3", "A3", 3, "a4", "A4", 4, "a5", "A5", 5)
                                    tempatParkir(350, 350, "b1", "B1", 6, "b2", "B2", 7, "b3", "B3", 8, "b4", "B4", 9, "b5", "B5", 10)
                                    tempatParkir(600, 350, "c1", "C1", 11, "c2", "C2", 12, "c3", "C3", 13, "c4", "C4", 14, "c5", "C5", 15)
                                    tempatParkir(660, 530, "d1", "D1", 16, "d2", "D2", 17, "d3", "D3", 18, "d4", "D4", 19, "d5", "D5", 20)

                                # Jika lokasi parkir pengguna <= 40 maka yang dipanggil adalah halaman lt2
                                elif dataset.lokasi_parkir[self.idx] <= 40:
                                    # Mengganti backgrond
                                    photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car2.jpg'))
                                    self.bgHome.configure(image=photo_bgHome)
                                    self.bgHome.image = photo_bgHome
                                    
                                    # Daftar nama kotak parkir di lt2
                                    self.dict_parkir = {21: 'E1', 22: 'E2', 23: 'E3', 24: 'E4', 25: 'E5',
                                                    26: 'F1', 27: 'F2', 28: 'F3', 29: 'F4', 30: 'F5',
                                                    31: 'G1', 32: 'G2', 33: 'G3', 34: 'G4', 35: 'G5',
                                                    36: 'H1', 37: 'H2', 38: 'H3', 39: 'H4', 40: 'H5'}

                                    # Fungsi untuk memanggil kotak parkir lt2
                                    tempatParkir(410, 530, "e1", "E1", 21, "e2", "E2", 22, "e3", "E3", 23, "e4", "E4", 24, "e5", "E5", 25)
                                    tempatParkir(350, 350, "f1", "F1", 26, "f2", "F2", 27, "f3", "F3", 28, "f4", "F4", 29, "f5", "F5", 30)
                                    tempatParkir(600, 350, "g1", "G1", 31, "g2", "G2", 32, "g3", "G3", 33, "g4", "G4", 34, "g5", "G5", 35)
                                    tempatParkir(660, 530, "h1", "H1", 36, "h2", "H2", 37, "h3", "H3", 38, "h4", "H4", 39, "h5", "H5", 40)

                                # Jika lokasi parkir pengguna > 40 maka yang dipanggil adalah halaman lt3
                                elif dataset.lokasi_parkir[self.idx] > 40:
                                    # Mengganti background
                                    photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car3.jpg'))
                                    self.bgHome.configure(image=photo_bgHome)
                                    self.bgHome.image = photo_bgHome
                                    
                                    # Daftar nama kotak parkir di lt3
                                    self.dict_parkir = {41: 'I1', 42: 'I2', 43: 'I3', 44: 'I4', 45: 'I5',
                                                    46: 'J1', 47: 'J2', 48: 'J3', 49: 'J4', 50: 'J5',
                                                    51: 'K1', 52: 'K2', 53: 'K3', 54: 'K4', 55: 'K5',
                                                    56: 'L1', 57: 'L2', 58: 'L3', 59: 'L4', 60: 'L5'}

                                    # Fungsi untuk memanggil kotak parkir lt3
                                    tempatParkir(410, 530, "i1", "I1", 41, "i2", "I2", 42, "i3", "I3", 43, "i4", "I4", 44, "i5", "I5", 45)
                                    tempatParkir(350, 350, "j1", "J1", 46, "j2", "J2", 47, "j3", "J3", 48, "j4", "J4", 49, "j5", "J5", 50)
                                    tempatParkir(600, 350, "k1", "K1", 51, "k2", "K2", 52, "k3", "K3", 53, "k4", "K4", 54, "k5", "K5", 55)
                                    tempatParkir(660, 530, "l1", "L1", 56, "l2", "L2", 57, "l3", "L3", 58, "l4", "L4", 59, "l5", "L5", 60)

                                # Pemberitahuan letak mobil
                                self.nomorMobil = Label(self.window, text=f"Halo {dataset['first name'][self.idx]}\nMobil anda berada pada nomor {self.dict_parkir[dataset.lokasi_parkir[self.idx]]}", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='center', anchor='center')
                                self.nomorMobil.place(x=490, y=600)

                                
                        # Transaction ===========================================================================================================================
                        def transaction():
                            # Membaca dataset
                            dataset = pd.read_csv("E-parking.csv")

                            # Jika pengguna belum parkir
                            if dataset.lokasi_parkir[self.idx] == 0:
                                showinfo("Notification", "Maaf, anda belum parkir\nsilahkan memilih tempat parkir terlebih dahulu")
                            
                            else:
                                def last():
                                    # Notifikasi pembayaran berhasil
                                    showinfo("Notification", "Pembayaran berhasil\nterimakasih sudah mempercayakan mobil anda pada kami")
                                    # Meriset data
                                    dataset.loc[self.idx, ["lokasi_parkir", "waktu_mulai", "waktu_selesai"]] = [0, "", ""]
                                    # Menyimpan dataset
                                    dataset.to_csv("E-parking.csv", index=False, header=True)
                                    # Memanggil halaman home
                                    self.homePage()

                                # Menyimpan waktu selesai parkir
                                dataset = pd.read_csv("E-parking.csv")
                                dataset.loc[self.idx, "waktu_selesai"] = datetime.now().strftime('%Y-%b-%d %H:%M:%S')
                                durasi, total = self.hitung_harga(dataset.waktu_mulai[self.idx], dataset.waktu_selesai[self.idx])
                                dataset.to_csv("E-parking.csv", index=False, header=True)

                                # Background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/transaksi page.jpg'))
                                self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                                self.bgHome.image = photo_bgHome
                                self.bgHome.place(x=0, y=0)

                                # Memanggil fungsi untuk merubah warna tombol pada sisi kiri menjadi abu abu
                                change_color()

                                # waktu dan jam
                                self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.datetime_label.place(x=1007, y=160)
                                self.update_datetime()

                                # label tanggal
                                self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.date_label.place(x=1007, y=140)

                                # Label waktu
                                self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.time_label.place(x=1154, y=140)

                                # Biodata akun di pojok kanan bawah
                                self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioFname.place(x=1100, y=503)

                                self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioLname.place(x=1100, y=546)

                                self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                                self.bioEmail.place(x=1100, y=583)

                                self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioRole.place(x=1100, y=622)


                                # Button transaksi kiri
                                dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi.jpg'))
                                self.payButton = Button(self.window, image=dollar_img,
                                                           relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43" ,command=transaction)
                                self.payButton.image = dollar_img
                                self.payButton.place(x=36, y=315)

                                # Nama pada bill
                                self.namaBill = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 11, "bold"), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w', height=1)
                                self.namaBill.place(x=410, y=381)

                                # Durasi pada bill
                                self.durasiParkir = Label(self.window, text=f"{durasi} Jam", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w')
                                self.durasiParkir.place(x=780, y=500)

                                # Total biaya pada bill
                                self.totalBill = Label(self.window, text=f"Rp.{total:,}", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w')
                                self.totalBill.place(x=780, y=540)

                                # Tombol dnone pada bill
                                done_img = ImageTk.PhotoImage(Image.open('Gambar/done.png'))
                                self.doneButton = Button(self.window, image=done_img,
                                                           relief=FLAT, height=40, width=100, bg="#2c2e43", activebackground="#2c2e43",
                                                          command=last)
                                self.doneButton.image = done_img
                                self.doneButton.place(x=570, y= 610)


                        # Background dashboard
                        photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/dashboard.jpg'))
                        self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                        self.bgHome.image = photo_bgHome
                        self.bgHome.place(x=0, y=0)

                        # waktu dan jam
                        self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                        self.datetime_label.place(x=1007, y=160)
                        self.update_datetime()

                        # label tanggal
                        self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                        self.date_label.place(x=1007, y=140)

                        # Label waktu
                        self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                        self.time_label.place(x=1154, y=140)

                        # Welcome teks
                        txt_welcome = """Temukan pengalaman parkir yang praktis, cepat, dan nyaman bersama kami.\nCari, pesan, dan bayar tempat parkir dengan mudah melalui aplikasi ini.\nTersedia peta interaktif untuk memudahkan Anda menemukan tempat parkir kosong.\nNikmati fitur notifikasi waktu parkir, pembayaran elektronik yang aman,\ndan pelacakan lokasi parkir. Dapatkan penawaran khusus dan diskon yang menarik.\nLayanan pelanggan kami siap membantu Anda. Selamat menikmati\nkemudahan parkir bersama Parking Apps!"""

                        self.welcome = Label(self.window, text=txt_welcome, font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='center', anchor='center', width=65)
                        self.welcome.place(x=330, y=390)

                        # Start Parking
                        start_img = ImageTk.PhotoImage(Image.open('Gambar/Start Parking.png'))
                        self.startButton = Button(self.window, image=start_img,
                                                   relief=FLAT, height=40, width=435, bg="#2c2e43", activebackground="#2c2e43",
                                                  command=startParking)
                        self.startButton.image = start_img
                        self.startButton.place(x=405, y= 580)

                        # Home button
                        home_img = ImageTk.PhotoImage(Image.open('Gambar/Home button1.jpg'))
                        self.homeButton = Button(self.window, image=home_img,
                                                   relief=FLAT, height=30, width=100, bg="#2c2e43", activebackground="#2c2e43")
                        self.homeButton.image = home_img
                        self.homeButton.place(x=40, y=160)

                        # Start Parking button
                        startIcon_img = ImageTk.PhotoImage(Image.open('Gambar/start parkir icon1.jpg'))
                        self.startIcon_Button = Button(self.window, image=startIcon_img,
                                                   relief=FLAT, height=30, width=150, bg="#2c2e43", activebackground="#2c2e43", command=startParking)
                        self.startIcon_Button.image = startIcon_img
                        self.startIcon_Button.place(x=30, y=210)

                        # Find my car button
                        find_img = ImageTk.PhotoImage(Image.open('Gambar/mobil1.jpg'))
                        self.findButton = Button(self.window, image=find_img,
                                                   relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", command=findMycar)
                        self.findButton.image = find_img
                        self.findButton.place(x=35, y=260)

                        # Button transaksi kiri
                        dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi1.jpg'))
                        self.payButton = Button(self.window, image=dollar_img,
                                                   relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43", command=transaction)
                        self.payButton.image = dollar_img
                        self.payButton.place(x=36, y=315)

                        # Nama pojok kanan atas
                        self.nickname = Label(self.window, text=f"Hi, {dataset['first name'][self.idx]}", bg="#24273a", fg="white",
                                        font=("yu gothic ui", 12, "bold"), width=50, anchor="e")
                        self.nickname.place(x=740, y=30)

                        # Biodata akun
                        self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                        self.bioFname.place(x=1100, y=503)

                        self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                        self.bioLname.place(x=1100, y=546)

                        self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                        self.bioEmail.place(x=1100, y=583)

                        self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                        self.bioRole.place(x=1100, y=622)
                        
                    
                    # Jika role pengguna merupakan Admin ========================================================================================================
                    else:
                        # Fungsi untuk mencari pengguna untuk menjalankan fiturnya
                        def cariUser():
                            # Halaman untuk mencari user berdasarkan email
                            def cariEmail():
                                # Jika entry tidak di isi
                                if not self.cariEmail_entry.get():
                                    showinfo("Notification", "Entry wajib di isi")
                                    return
                                    
                                # Melakukan pengurutan dan pencarian index
                                dataset = pd.read_csv("E-parking.csv")
                                dataset = self.bubble_sort(dataset, 'email')
                                dataset.reset_index(drop=True, inplace=True)
                                dataset.to_csv("E-parking.csv", index=False, header=True)
                                self.idx1 = self.binary_search("E-parking.csv", self.cariEmail_entry.get(), "email")
                                
                                # Jika index kosong atau index yg dicari sama dengan index pengguna(admin)
                                if self.idx1 == None or self.idx1 == self.idx:
                                    showinfo("Notification", "E-mail yang anda cari tidak ada")
                            
                                # Jika email ditemukan
                                else:
                                    showinfo("Notification", "Berhasil, silahkan gunakan fitur yang telah tersedia")
                                    
                                    # Mengembalikan tombol agar bisa ditekan kembali
                                    self.findButton.configure(state=NORMAL)
                                    self.payButton.configure(state=NORMAL)
                                
                            # Background dashboard
                            photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/dashboard.jpg'))
                            self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                            self.bgHome.image = photo_bgHome
                            self.bgHome.place(x=0, y=0)

                            # waktu dan jam
                            self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.datetime_label.place(x=1007, y=160)
                            self.update_datetime()

                            # label tanggal
                            self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                            self.date_label.place(x=1007, y=140)

                            # Label waktu
                            self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                            self.time_label.place(x=1154, y=140)

                            # Welcome teks
                            txt_welcome = f"Selamat datang {dataset['first name'][self.idx]}\nsilahkan isi e-mail yang ingin dicari pada kolom dibawh ini"

                            self.welcome = Label(self.window, text=txt_welcome, font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='center', anchor='center', width=65)
                            self.welcome.place(x=330, y=390)
                            
                            # Entry cari email
                            self.cariEmail_entry = Entry(self.window, highlightthickness=0, relief=FLAT, bg="#5d5f70", fg="white",
                                                        font=("yu gothic ui ", 12, "bold"), width=47)
                            self.cariEmail_entry.place(x=413, y=460)

                            # Home button
                            home_img = ImageTk.PhotoImage(Image.open('Gambar/Home button1.jpg'))
                            self.homeButton = Button(self.window, image=home_img,
                                                       relief=FLAT, height=30, width=100, bg="#2c2e43", activebackground="#2c2e43")
                            self.homeButton.image = home_img
                            self.homeButton.place(x=40, y=160)

                            # Find my car button
                            car_img = ImageTk.PhotoImage(Image.open('Gambar/mobil1.jpg'))
                            self.findButton = Button(self.window, image=car_img,
                                                       relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", state=DISABLED, command=findMycar)
                            self.findButton.image =car_img
                            self.findButton.place(x=35, y=210)

                            # Button transaksi kiri
                            dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi1.jpg'))
                            self.payButton = Button(self.window, image=dollar_img,
                                                       relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43", state=DISABLED, command=transaction)
                            self.payButton.image = dollar_img
                            self.payButton.place(x=36, y=260)

                            # Nama pojok kanan atas
                            self.nickname = Label(self.window, text=f"Hi, {dataset['first name'][self.idx]}", bg="#24273a", fg="white",
                                            font=("yu gothic ui", 12, "bold"), width=50, anchor="e")
                            self.nickname.place(x=740, y=30)

                            # Biodata akun
                            self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioFname.place(x=1100, y=503)

                            self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioLname.place(x=1100, y=546)

                            self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                            self.bioEmail.place(x=1100, y=583)

                            self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                            self.bioRole.place(x=1100, y=622)
                            
                            # Tombol done
                            done_img = ImageTk.PhotoImage(Image.open('Gambar/done.png'))
                            self.doneButton = Button(self.window, image=done_img,
                                                       relief=FLAT, height=40, width=100, bg="#2c2e43", activebackground="#2c2e43",
                                                      command=cariEmail)
                            self.doneButton.image = done_img
                            self.doneButton.place(x=570, y= 490)
                            
                        # fing my car ======================================================================================================
                        def findMycar():
                            dataset = pd.read_csv("E-parking.csv")

                            # Jika user yang dicari belum parkir
                            if dataset.lokasi_parkir[self.idx1] == 0:
                                # Peringatan belum parkir
                                showinfo("Notification", "Maaf, anda belum parkir\nsilahkan memilih tempat parkir terlebih dahulu")
                            else:
                                # Background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car.jpg'))
                                self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                                self.bgHome.image = photo_bgHome
                                self.bgHome.place(x=0, y=0)

                                # memanggil fungsi untuk mengubah warna tombol di sisi kiri menajadi abu abu 
                                change_color()

                                # Find my car button
                                car_img = ImageTk.PhotoImage(Image.open('Gambar/mobil.jpg'))
                                self.findButton = Button(self.window, image=car_img,
                                                           relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", command=findMycar)
                                self.findButton.image = car_img
                                self.findButton.place(x=35, y=210)

                                # waktu dan jam
                                self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.datetime_label.place(x=1007, y=160)
                                self.update_datetime()

                                # label tanggal
                                self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.date_label.place(x=1007, y=140)

                                # Label waktu
                                self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.time_label.place(x=1154, y=140)
                                
                                # Biodata akun
                                self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioFname.place(x=1100, y=503)
                        
                                self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioLname.place(x=1100, y=546)

                                self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                                self.bioEmail.place(x=1100, y=583)

                                self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioRole.place(x=1100, y=622)

                                # Kotak kotak
                                # Kotak kotak
                                def tempatParkir(locx, locy, namaVar1, teks1, value1, namaVar2, teks2, value2, namaVar3, teks3, value3, namaVar4, teks4, value4, namaVar5, teks5, value5):

                                    # Memeriksa tempat parkir yang sudah digunakan
                                    self.ketersediaan_parkir = [i for i in dataset.lokasi_parkir]

                                    # Membuat dan menempatkan tombol pertama
                                    setattr(self, namaVar1, Button(self.window, text=teks1, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar1).place(x=locx, y=locy)

                                    # Memeriksa apakah tempat parkir pertama sudah digunakan
                                    if value1 in self.ketersediaan_parkir:
                                        # Jika value 1 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx1] == value1:
                                            getattr(self, namaVar1)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar1)["bg"] = "#4a495b"
                                            getattr(self, namaVar1)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol kedua
                                    setattr(self, namaVar2, Button(self.window, text=teks2, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar2).place(x=locx+35, y=locy)

                                    # Memeriksa apakah tempat parkir kedua sudah digunakan
                                    if value2 in self.ketersediaan_parkir:
                                        # Jika value 2 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx1] == value2:
                                            getattr(self, namaVar2)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar2)["bg"] = "#4a495b"
                                            getattr(self, namaVar2)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol ketiga
                                    setattr(self, namaVar3, Button(self.window, text=teks3, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar3).place(x=locx+70, y=locy)

                                    # Memeriksa apakah tempat parkir ketiga sudah digunakan
                                    if value3 in self.ketersediaan_parkir:
                                        # Jika value 3 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx1] == value3:
                                            getattr(self, namaVar3)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar3)["bg"] = "#4a495b"
                                            getattr(self, namaVar3)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol keempat
                                    setattr(self, namaVar4, Button(self.window, text=teks4, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar4).place(x=locx+105, y=locy)

                                    # Memeriksa apakah tempat parkir keempat sudah digunakan
                                    if value4 in self.ketersediaan_parkir:
                                        # Jika value 4 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx1] == value4:
                                            getattr(self, namaVar4)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar4)["bg"] = "#4a495b"
                                            getattr(self, namaVar4)["state"] = DISABLED

                                    # Membuat dan menempatkan tombol kelima
                                    setattr(self, namaVar5, Button(self.window, text=teks5, relief=FLAT, height=2, width=2, bg="#f4ce21",
                                                   activebackground="#4a495b"))
                                    getattr(self, namaVar5).place(x=locx+140, y=locy)

                                    # Memeriksa apakah tempat parkir keempat sudah digunakan
                                    if value5 in self.ketersediaan_parkir:
                                        # Jika value 5 = lokasi parkir pengguna, maka ubah bacckground jadi hijau
                                        if dataset.lokasi_parkir[self.idx1] == value5:
                                            getattr(self, namaVar5)["bg"] = "#16FF00"
                                        else:
                                            getattr(self, namaVar5)["bg"] = "#4a495b"
                                            getattr(self, namaVar5)["state"] = DISABLED    

                                # Jika lokasi parkir pengguna yang dicari <= 20
                                if dataset.lokasi_parkir[self.idx1] <= 20:
                                    # Daftar nama kolom parkir
                                    self.dict_parkir = {1: 'A1', 2: 'A2', 3: 'A3', 4: 'A4', 5: 'A5',
                                                     6: 'B1', 7: 'B2', 8: 'B3', 9: 'B4', 10: 'B5',
                                                     11: 'C1', 12: 'C2', 13: 'C3', 14: 'C4', 15: 'C5',
                                                     16: 'D1', 17: 'D2', 18: 'D3', 19: 'D4', 20: 'D5'}

                                    tempatParkir(410, 530, "a1", "A1", 1, "a2", "A2", 2, "a3", "A3", 3, "a4", "A4", 4, "a5", "A5", 5)
                                    tempatParkir(350, 350, "b1", "B1", 6, "b2", "B2", 7, "b3", "B3", 8, "b4", "B4", 9, "b5", "B5", 10)
                                    tempatParkir(600, 350, "c1", "C1", 11, "c2", "C2", 12, "c3", "C3", 13, "c4", "C4", 14, "c5", "C5", 15)
                                    tempatParkir(660, 530, "d1", "D1", 16, "d2", "D2", 17, "d3", "D3", 18, "d4", "D4", 19, "d5", "D5", 20)

                                # Jika lokasi parkir pengguna yang dicari <= 40
                                elif dataset.lokasi_parkir[self.idx1] <= 40:
                                    # Mengganti background
                                    photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car2.jpg'))
                                    self.bgHome.configure(image=photo_bgHome)
                                    self.bgHome.image = photo_bgHome
                                    
                                    # Dictionary daftar tempat parkir pada lantai 2
                                    self.dict_parkir = {21: 'E1', 22: 'E2', 23: 'E3', 24: 'E4', 25: 'E5',
                                                    26: 'F1', 27: 'F2', 28: 'F3', 29: 'F4', 30: 'F5',
                                                    31: 'G1', 32: 'G2', 33: 'G3', 34: 'G4', 35: 'G5',
                                                    36: 'H1', 37: 'H2', 38: 'H3', 39: 'H4', 40: 'H5'}

                                    tempatParkir(410, 530, "e1", "E1", 21, "e2", "E2", 22, "e3", "E3", 23, "e4", "E4", 24, "e5", "E5", 25)
                                    tempatParkir(350, 350, "f1", "F1", 26, "f2", "F2", 27, "f3", "F3", 28, "f4", "F4", 29, "f5", "F5", 30)
                                    tempatParkir(600, 350, "g1", "G1", 31, "g2", "G2", 32, "g3", "G3", 33, "g4", "G4", 34, "g5", "G5", 35)
                                    tempatParkir(660, 530, "h1", "H1", 36, "h2", "H2", 37, "h3", "H3", 38, "h4", "H4", 39, "h5", "H5", 40)

                                # Jika lokasi parkir pengguna yang dicari > 40
                                elif dataset.lokasi_parkir[self.idx1] > 40:
                                    # Mengganti background
                                    photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/find my car3.jpg'))
                                    self.bgHome.configure(image=photo_bgHome)
                                    self.bgHome.image = photo_bgHome
                                    
                                    # Dictionary daftar tempat parkir pada lantai 2
                                    self.dict_parkir = {41: 'I1', 42: 'I2', 43: 'I3', 44: 'I4', 45: 'I5',
                                                    46: 'J1', 47: 'J2', 48: 'J3', 49: 'J4', 50: 'J5',
                                                    51: 'K1', 52: 'K2', 53: 'K3', 54: 'K4', 55: 'K5',
                                                    56: 'L1', 57: 'L2', 58: 'L3', 59: 'L4', 60: 'L5'}

                                    tempatParkir(410, 530, "i1", "I1", 41, "i2", "I2", 42, "i3", "I3", 43, "i4", "I4", 44, "i5", "I5", 45)
                                    tempatParkir(350, 350, "j1", "J1", 46, "j2", "J2", 47, "j3", "J3", 48, "j4", "J4", 49, "j5", "J5", 50)
                                    tempatParkir(600, 350, "k1", "K1", 51, "k2", "K2", 52, "k3", "K3", 53, "k4", "K4", 54, "k5", "K5", 55)
                                    tempatParkir(660, 530, "l1", "L1", 56, "l2", "L2", 57, "l3", "L3", 58, "l4", "L4", 59, "l5", "L5", 60)

                                # Pemberitahuan nomor tempat parkir
                                self.nomorMobil = Label(self.window, text=f"Akun dengan nama pemilik {dataset['first name'][self.idx1]}\nparkir pada nomor {self.dict_parkir[dataset.lokasi_parkir[self.idx1]]}", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='center', anchor='center')
                                self.nomorMobil.place(x=520, y=600)

                                
                        # Transaction =================================================================================================
                        def transaction():
                            # Membaca dataset
                            dataset = pd.read_csv("E-parking.csv")
                            
                            # Jika belum parkir
                            if dataset.lokasi_parkir[self.idx1] == 0:
                                showinfo("Notification", "Maaf, anda belum parkir\nsilahkan memilih tempat parkir terlebih dahulu")
                            else:
                                # Fungsi untuk menyelesaikan transaksi dan mereset data pada kolom lokasi parkir, waktu mulai dan selesai
                                def last():
                                    showinfo("Notification", "Pembayaran berhasil\nterimakasih sudah mempercayakan mobil anda pada kami")
                                    dataset = pd.read_csv("E-parking.csv")
                                    dataset.loc[self.idx1, ["lokasi_parkir", "waktu_mulai", "waktu_selesai"]] = [0, "", ""]
                                    dataset.to_csv("E-parking.csv", index=False, header=True)
                                    self.homePage()

                                # Menghitung durasi dan total biaya
                                dataset = pd.read_csv("E-parking.csv")
                                dataset.loc[self.idx1, "waktu_selesai"] = datetime.now().strftime('%Y-%b-%d %H:%M:%S')
                                durasi, total = self.hitung_harga(dataset.waktu_mulai[self.idx1], dataset.waktu_selesai[self.idx1])
                                dataset.to_csv("E-parking.csv", index=False, header=True)
                                
                                # Background
                                photo_bgHome = ImageTk.PhotoImage(Image.open('Gambar/transaksi page.jpg'))
                                self.bgHome = Label(self.window, image=photo_bgHome, width=1280, height=730)
                                self.bgHome.image = photo_bgHome
                                self.bgHome.place(x=0, y=0)

                                change_color()

                                # waktu dan jam
                                self.datetime_label = Label(self.window, text='', font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.datetime_label.place(x=1007, y=160)
                                self.update_datetime()

                                # label tanggal
                                self.date_label = Label(self.window, text='Date', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.date_label.place(x=1007, y=140)

                                # Label waktu
                                self.time_label = Label(self.window, text='Time', font=("yu gothic ui", 12), bg="#2c2e43", fg="#ffc800")
                                self.time_label.place(x=1154, y=140)

                                # Biodata akun
                                self.bioFname = Label(self.window, text=dataset["first name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioFname.place(x=1100, y=503)

                                self.bioLname = Label(self.window, text=dataset["last name"][self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioLname.place(x=1100, y=546)

                                self.bioEmail = Label(self.window, text=dataset.email[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", width=14, anchor="w")
                                self.bioEmail.place(x=1100, y=583)

                                self.bioRole = Label(self.window, text=dataset.role[self.idx], font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9")
                                self.bioRole.place(x=1100, y=622)


                                # Button transaksi kiri
                                dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi.jpg'))
                                self.payButton = Button(self.window, image=dollar_img,
                                                           relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43" ,command=transaction)
                                self.payButton.image = dollar_img
                                self.payButton.place(x=36, y=260)
                                
                                # Nama pada Bill
                                self.namaBill = Label(self.window, text=dataset["first name"][self.idx1], font=("yu gothic ui", 11, "bold"), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w', height=1)
                                self.namaBill.place(x=410, y=381)

                                # Durasi jam pada bill
                                self.labelJam = Label(self.window, text=f"{durasi} Jam", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w')
                                self.labelJam.place(x=780, y=500)

                                # Total biaya pada bill
                                self.labelTotal = Label(self.window, text=f"Rp.{total:,}", font=("yu gothic ui", 12), bg="#2c2e43", fg="#B2B1B9", justify='left', anchor='w')
                                self.labelTotal.place(x=780, y=540)

                                # Tombol done
                                done_img = ImageTk.PhotoImage(Image.open('Gambar/done.png'))
                                self.doneButton = Button(self.window, image=done_img,
                                                           relief=FLAT, height=40, width=100, bg="#2c2e43", activebackground="#2c2e43",
                                                          command=last)
                                self.doneButton.image = done_img
                                self.doneButton.place(x=570, y= 610)
                                
                    
                        cariUser()
                        
                        def change_color():
                            # Nama pojok kanan atas
                            self.nickname = Label(self.window, text=f"Hi, {dataset['first name'][self.idx]}", bg="#24273a", fg="white",
                                            font=("yu gothic ui", 12, "bold"), width=50, anchor="e")
                            self.nickname.place(x=740, y=30)

                            # Home button
                            home_img = ImageTk.PhotoImage(Image.open('Gambar/Home button.jpg'))
                            self.homeButton = Button(self.window, image=home_img,
                                                   relief=FLAT, height=30, width=100, activebackground="#2c2e43", command=cariUser)
                            self.homeButton.image = home_img
                            self.homeButton.place(x=40, y=160)

                            # Find my car button kiri
                            car_img = ImageTk.PhotoImage(Image.open('Gambar/mobil1.jpg'))
                            self.findButton = Button(self.window, image=car_img,
                                                   relief=FLAT, height=30, width=140, bg="#2c2e43", activebackground="#2c2e43", command=findMycar)
                            self.findButton.image = car_img
                            self.findButton.place(x=35, y=210)

                            # Button transaksi kiri
                            dollar_img = ImageTk.PhotoImage(Image.open('Gambar/transaksi1.jpg'))
                            self.payButton = Button(self.window, image=dollar_img,
                                                   relief=FLAT, height=33, width=140, bg="#2c2e43", activebackground="#2c2e43", command=transaction)
                            self.payButton.image = dollar_img
                            self.payButton.place(x=36, y=260)
                        
                # Peringatan email atau sandi salah    
                else:
                    destroy_check_log()
                    self.emailSandi = Label(self.window, background="#2c2e43", text='Incorrect email or password',
                                    fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                    self.emailSandi.place(x=200, y=490)
                    
            # Peringatan email atau sandi salah
            else:
                destroy_check_log()
                self.emailSandi = Label(self.window, background="#2c2e43", text='Incorrect email or password',
                                fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
                self.emailSandi.place(x=200, y=490)
                    
        # Peringatan semua entry harus di isi
        else:
            destroy_check_log()
            self.wajibMengisi = Label(self.window, background="#2c2e43", text='All inputs are required',
                            fg='red', font=("yu gothic ui", 12), relief=FLAT, width=30, height=1)
            self.wajibMengisi.place(x=200, y=490)
    
    def update_datetime(self):
        # Waktu dan jam saat ini
        current_datetime = datetime.now().strftime('%d-%b-%Y \t %H:%M:%S')
        # Mengganti text pada label date time dengan waktu dan jam saat ini
        self.datetime_label.config(text=current_datetime)
        # Update setiap 1000 milidetik (1 detik)
        self.datetime_label.after(1000, self.update_datetime)  
        
    def hitung_harga(self, mulai, waktu_selesai, tarif_per_jam=20000):
        
        format_waktu = '%Y-%b-%d %H:%M:%S'
        waktu_mulai = datetime.strptime(mulai, format_waktu)
        waktu_selesai = datetime.strptime(waktu_selesai, format_waktu)
        durasi = waktu_selesai - waktu_mulai
        # Dibagi 3600 karena 1 jam = 3600 detik
        durasi_jam = int(durasi.total_seconds() / 3600)
        harga_total = tarif_per_jam + 5000 * durasi_jam
        return durasi_jam, harga_total
        
        
    def bubble_sort(self, df, column):
        n = len(df)
        for i in range(n):
            flag = True
            for j in range(0, n-i-1):
                if df.loc[j, column] > df.loc[j+1, column]:
                    df.iloc[j], df.iloc[j+1] = df.iloc[j+1], df.iloc[j]
                    flag = False
            if flag:
                break

        return df
    
    def binary_search(self, csv_file, search_value, search_column):
        data = pd.read_csv(csv_file)
        data_sorted = data.sort_values(search_column)

        left = 0
        right = len(data_sorted) - 1

        while left <= right:
            mid = (left + right) // 2
            mid_value = data_sorted.iloc[mid][search_column]

            if mid_value == search_value:
                return mid
            elif mid_value < search_value:
                left = mid + 1
            else:
                right = mid - 1

        return None

    
window = Tk()
root = E_parking(window)
window.mainloop()  
```

```python
from datetime import datetime, timedelta

def hitung_harga(tarif_per_jam, waktu_mulai, waktu_selesai):
    format_waktu = '%Y-%b-%d %H:%M:%S'
    waktu_mulai = datetime.strptime(waktu_mulai, format_waktu)
    waktu_selesai = datetime.strptime(waktu_selesai, format_waktu)
    durasi = waktu_selesai - waktu_mulai
    durasi_jam = durasi.total_seconds() / 3600
    harga_total = tarif_per_jam * durasi_jam
    return harga_total

# Contoh penggunaan fungsi hitung_harga
tarif_per_jam = 100
waktu_mulai = '2023-May-28 09:30:00'
waktu_selesai = '2023-May-28 12:30:00'
harga = hitung_harga(tarif_per_jam, waktu_mulai, waktu_selesai)
print(f'Harga total: {harga} USD')
```
